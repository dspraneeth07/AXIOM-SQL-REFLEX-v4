```python
import json
from pathlib import Path

bird_path = Path("../data/bird")

# Check base folder
if not bird_path.exists():
    raise FileNotFoundError("BIRD folder not found")

# Prefer dev.json (public)
dev_file = bird_path / "dev.json"

if not dev_file.exists():
    raise FileNotFoundError("BIRD dev.json not found (this is expected in public BIRD)")

with open(dev_file, "r", encoding="utf-8") as f:
    bird_data = json.load(f)

print("✅ BIRD loaded successfully")
print("BIRD samples (dev):", len(bird_data))
print("Sample question:", bird_data[0].get("question"))
```

```python
import duckdb
from pathlib import Path

# --- Path to Spider SQLite DB ---
db_path = Path("../data/spider/database/academic/academic.sqlite").resolve()

if not db_path.exists():
    raise FileNotFoundError(f"SQLite DB not found at {db_path}")

# --- Connect to DuckDB ---
con = duckdb.connect(database=":memory:")

# --- Install & load sqlite extension ---
con.execute("INSTALL sqlite;")
con.execute("LOAD sqlite;")

# --- Attach Spider DB ---
con.execute(f"""
ATTACH DATABASE '{db_path}' AS spider_db (TYPE sqlite);
""")

# --- Correct DuckDB-native way to list tables ---
tables = con.execute("""
SELECT table_name
FROM duckdb_tables()
WHERE database_name = 'spider_db';
""").fetchall()

print("✅ Spider DB attached successfully")
print("Tables:")
for t in tables:
    print(" -", t[0])

```
